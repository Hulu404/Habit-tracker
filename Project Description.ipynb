# Трекер Привычек


Суть бота в том, что в него будет загружаться Road Map для освоения какого-либо навыка. Он будет каждый день прислылать статистику, на сколько удалось пройти путь к освоению навыка.

Его функционал:

    Ежедневно присылать сообщения с напоминанием об пути, список дел на сегодняшний день.

    По средствам сообщений будет принимать результаты и заполнять их в шкалы прогрресса.

    Принимать RoadMap и разбивать её по дням, разделяя его на блоки обучения



Первый сприт

2 дня

На этом спринте задачи:

    Создание шкалы по сообщению

    Добавление XP в шкалу

    Добавление RoadMap и разбиение его на блоки
